### Loading a Dataset with pandas

```python
import pandas as pd

# Load the brain size dataset from the current directory (notebooks)
data = pd.read_csv(r'brain_size (2).csv', sep=';', na_values='.')

# Display the first few rows to check it loaded correctly
data.head()
```

###  Generating and Transforming Data with NumPy

In this section, we use NumPy to create a sequence of evenly spaced numbers between -6 and 6. We then compute the sine and cosine values for each of these points. These arrays will be useful for plotting or further analysis of trigonometric functions.


```python
import numpy as np

# Generate 20 evenly spaced numbers between -6 and 6 (inclusive)
t = np.linspace(-6, 6, 20)

# Calculate the sine of each value in the array 't'
sin_t = np.sin(t)

# Calculate the cosine of each value in the array 't'
cos_t = np.cos(t)
```

###  Creating a DataFrame to Organize Computed Values

Here, we organize the generated values into a pandas DataFrame. The DataFrame includes the original sequence `t` alongside its corresponding sine and cosine values. This tabular structure makes it easier to analyze, visualize, and manipulate the data.


```python
import pandas as pd

# Create a DataFrame with three columns:
# - 't': the sequence of values from -6 to 6
# - 'sin': the sine of each 't' value
# - 'cos': the cosine of each 't' value
df = pd.DataFrame({'t': t, 'sin': sin_t, 'cos': cos_t})

# Display the DataFrame to verify its contents
df
```

###  Manipulating the Dataset Structure

In this section, we inspect the shape and columns of the dataset to understand its dimensions and structure. We also demonstrate how to access a specific column by its name — here, the `Gender` column — which is useful for analysis and filtering.


```python
# Display the shape of the dataset: number of rows and columns
print(data.shape)  # Expected output: (40, 8) - 40 rows and 8 columns

# List all column names in the dataset
print(data.columns)  # Shows the names of all columns

# Access and print the 'Gender' column from the dataset
print(data['Gender'])  # Columns can be accessed by their names like this

# Filter rows where Gender is 'Female'
# Select the 'VIQ' column from these filtered rows
# Calculate and return the mean of the selected 'VIQ' values
female_viq_mean = data[data['Gender'] == 'Female']['VIQ'].mean()

# Display the calculated average Verbal IQ for females
print(female_viq_mean)
```

### Grouping Data and Calculating Mean VIQ by Gender

This section demonstrates how to group the dataset by the `Gender` column and calculate the average Verbal IQ (`VIQ`) for each gender group. Grouping is a powerful way to perform aggregate calculations on subsets of your data.


```python
# Group the dataset by 'Gender'
groupby_gender = data.groupby('Gender')

# Loop through each gender group and corresponding 'VIQ' values
for gender, values in groupby_gender['VIQ']:
    # Calculate the mean VIQ for the current gender group and print the result
    print((gender, values.mean()))
```

###  Calculating Mean Values for All Numeric Columns by Gender

Using the groupby object, this command calculates the mean of all numeric columns in the dataset, grouped by `Gender`. This provides a quick summary of average values across multiple variables for each gender group.


```python
# Calculate the mean for all numeric columns grouped by 'Gender'
gender_means = groupby_gender.mean()

# Display the resulting DataFrame with average values for each gender
print(gender_means)
```

###  Hypothesis Testing: Comparing Two Groups

In this section, we will perform simple statistical tests to compare two groups. Hypothesis testing allows us to determine if observed differences between groups are statistically significant or likely due to chance.

We will use the `scipy.stats` sub-module from the SciPy library, which provides a wide range of statistical functions including t-tests, chi-square tests, and more.


```python
# Import the stats sub-module from scipy, which contains statistical functions for hypothesis testing
from scipy import stats
```

```python
# Perform a one-sample t-test to check if the mean VIQ is significantly different from 0
# H0 (null hypothesis): The mean VIQ = 0
# H1 (alternative hypothesis): The mean VIQ ≠ 0
stats.ttest_1samp(data['VIQ'], 0)
```

```python
# Extract VIQ scores for females
female_viq = data[data['Gender'] == 'Female']['VIQ']

# Extract VIQ scores for males
male_viq = data[data['Gender'] == 'Male']['VIQ']

# Perform an independent two-sample t-test to compare VIQ between genders
# H0: mean(Female VIQ) == mean(Male VIQ)
# H1: mean(Female VIQ) ≠ mean(Male VIQ)
stats.ttest_ind(female_viq, male_viq)
```

###  Comparing Full-Scale IQ (FSIQ) and Performance IQ (PIQ)

In this analysis, we use an **independent two-sample t-test** (`ttest_ind`) to statistically compare the **FSIQ** (Full-Scale IQ) and **PIQ** (Performance IQ) scores across all participants.

- **Null Hypothesis (H₀):** The mean FSIQ is equal to the mean PIQ.
- **Alternative Hypothesis (H₁):** The mean FSIQ is significantly different from the mean PIQ.

Although both variables come from the same individuals (and a paired t-test might seem more appropriate), here we demonstrate the use of the independent t-test to understand its behavior.


```python
# Perform an independent two-sample t-test to compare FSIQ and PIQ scores
# This checks if the average FSIQ is statistically different from the average PIQ
stats.ttest_ind(data['FSIQ'], data['PIQ'])
```

```python
# Import the necessary libraries
import pandas as pd
from scipy import stats

# Load the dataset using the correct full path
# Note: We're using a raw string (r'') to avoid issues with special characters like parentheses and spaces
data = pd.read_csv(r'/users/PAS2797/kehindesoetan/Stats_Assignment4/notebooks/brain_size (2).csv', sep=';', na_values='.')

# Perform a one-sample t-test on the difference between FSIQ and PIQ
# Null Hypothesis: The mean difference between FSIQ and PIQ is zero
# Alternative Hypothesis: The mean difference is not zero
result = stats.ttest_1samp(data['FSIQ'] - data['PIQ'], 0)

# Output the t-test results
print(result)
```

### Wilcoxon Signed-Rank Test:
Used to compare two related (paired) samples — in this case, Full Scale IQ (FSIQ) and Performance IQ (PIQ). Unlike the paired t-test, which assumes the data is normally distributed, the Wilcoxon test is a non-parametric alternative that tests whether the median difference between paired observations is zero.

Null Hypothesis (H₀): The median of the differences between FSIQ and PIQ is zero.

Alternative Hypothesis (H₁): The median difference is not zero.

Use this test when the distribution of differences is not symmetric or when you're unsure about the normality of the data.

```python
# Perform a Wilcoxon signed-rank test to compare paired FSIQ and PIQ scores
# This non-parametric test is used when the assumptions of the paired t-test (like normality) are not met
# It tests whether the median difference between the paired scores is significantly different from zero
stats.wilcoxon(data['FSIQ'], data['PIQ'])
```

### Creating a Simple Linear Dataset:

x: A range of 20 values between -5 and 5, representing the independent variable.

np.random.seed(1): Ensures that the random numbers generated are reproducible every time the script runs.

y: A linear function of x with added normally distributed random noise, simulating real-world data.

Formula used: y = -5 + 3x + noise

Noise is scaled by a factor of 4, introducing variability.

Finally, the code wraps x and y into a pandas.DataFrame called data, which can be easily used for modeling, visualization, or statistical analysis.

```python
import pandas as pd  # Import pandas library with alias 'pd'
import numpy as np   # Import numpy for numerical operations

x = np.linspace(-5, 5, 20)
np.random.seed(1)
y = -5 + 3*x + 4 * np.random.normal(size=x.shape)

# Use pd.DataFrame (since we imported pandas as pd) to create the DataFrame
data = pd.DataFrame({'x': x, 'y': y})
```

```python
# Import the ordinary least squares (OLS) regression function from the statsmodels formula API
from statsmodels.formula.api import ols

# Fit a linear regression model with 'y' as the dependent variable and 'x' as the independent variable
# The formula 'y ~ x' specifies the relationship to model
# The .fit() method estimates the regression coefficients (intercept and slope)
model = ols("y ~ x", data).fit()

# Print the summary of the regression results, including coefficients, statistics, and diagnostics
print(model.summary())
```

### Categorical Variables: Comparing Groups or Multiple Categories
In this section, we explore how to analyze and compare data across different categories or groups. Categorical variables represent discrete groups—like gender, treatment types, or categories—and often require different statistical approaches than continuous data.

We’ll look at ways to:

Select data by group,

Calculate group-specific statistics,

Perform hypothesis tests to compare groups,

And visualize differences between categories.

This approach helps uncover meaningful patterns and differences between distinct groups in your dataset.





```python
import pandas as pd  # Import pandas library for data manipulation

# Load the brain size dataset from the specified file path
# Use semicolon as the separator and treat '.' as missing values (NaN)
data = pd.read_csv(r'/users/PAS2797/kehindesoetan/Stats_Assignment4/notebooks/brain_size (2).csv', sep=';', na_values='.')
```

```python
# Fit an ordinary least squares (OLS) regression model to predict VIQ based on Gender
# The formula 'VIQ ~ Gender + 1' means VIQ is modeled as a function of Gender with an intercept
# This allows us to examine the effect of Gender on VIQ scores
# The .fit() method estimates the regression coefficients
model = ols("VIQ ~ Gender + 1", data).fit()

# Print a detailed summary of the regression results, including coefficients, p-values, and model diagnostics
print(model.summary())
```

### Specifying Categorical Variables in the Regression Model
In this model, we predict Verbal IQ (VIQ) using Gender as a categorical predictor. The key part here is C(Gender), which explicitly instructs statsmodels to treat Gender as a categorical variable rather than a continuous one.

This means the model will estimate separate coefficients for each gender group.

One gender category acts as the reference group, and other categories’ effects are interpreted relative to this baseline.

This approach is essential when working with categorical data in regression to ensure correct interpretation and statistical testing.

```python
# Fit an OLS regression model predicting VIQ based on Gender as a categorical variable
# Using 'C(Gender)' explicitly tells statsmodels to treat Gender as a categorical predictor
# This creates separate coefficients for each gender category, with one serving as the baseline (reference)
# The .fit() method estimates the regression coefficients for the model
model = ols('VIQ ~ C(Gender)', data).fit()
```

### Reshaping Data for Comparison: Long Format and Regression
Here, we reshape the IQ data to compare Full-Scale IQ (FSIQ) and Performance IQ (PIQ) scores using regression analysis:

We create two separate DataFrames, one for FSIQ and one for PIQ, each with an 'iq' column for scores and a 'type' column labeling the IQ type.

These DataFrames are concatenated into a long format DataFrame, where all IQ scores are stacked vertically, making it easier to analyze differences between groups.

Using this long format, we fit an OLS regression model to predict IQ scores based on their type (fsiq vs. piq).

This approach tests whether there is a statistically significant difference in mean IQ scores between the two types.

```python
# Create a new DataFrame for Full-Scale IQ (FSIQ) scores with a column 'type' labeled 'fsiq'
data_fisq = pd.DataFrame({'iq': data['FSIQ'], 'type': 'fsiq'})

# Create a new DataFrame for Performance IQ (PIQ) scores with a column 'type' labeled 'piq'
data_piq = pd.DataFrame({'iq': data['PIQ'], 'type': 'piq'})

# Concatenate the two DataFrames vertically to create a 'long' format DataFrame
# This format stacks the IQ scores together with their corresponding type labels
data_long = pd.concat((data_fisq, data_piq))

# Print the combined DataFrame to verify its structure
print(data_long)

# Fit an OLS regression model predicting IQ score ('iq') based on the IQ type ('type')
# This model tests whether the mean IQ scores differ significantly between FSIQ and PIQ
model = ols("iq ~ type", data_long).fit()

# Display a detailed summary of the regression results, including coefficients and p-values
print(model.summary())
```

### Comparing IQ Types Using Independent Two-Sample t-Test
To determine whether there is a statistically significant difference between the mean Full-Scale IQ (FSIQ) and Performance IQ (PIQ) scores, we use an independent two-sample t-test:

```python
# Perform an independent two-sample t-test comparing FSIQ and PIQ scores
# This test checks whether the means of the two independent groups (FSIQ and PIQ) are significantly different
# Null Hypothesis (H0): The mean FSIQ and PIQ scores are equal
# Alternative Hypothesis (H1): The mean FSIQ and PIQ scores are different
stats.ttest_ind(data['FSIQ'], data['PIQ'])
```

### Multiple Regression: Modeling Relationships with Multiple Predictors
Multiple regression allows us to model the relationship between a dependent variable and two or more independent variables simultaneously. This approach helps us understand how each predictor uniquely contributes to explaining the variability in the outcome, controlling for the presence of other variables.

```python
import pandas as pd
import seaborn as sns
from statsmodels.formula.api import ols

# Load the iris dataset directly from seaborn's built-in datasets
data = sns.load_dataset('iris')

# Fit a multiple regression model predicting sepal_width based on species (categorical) and petal_length (continuous)
model = ols('sepal_width ~ species + petal_length', data).fit()

# Print the detailed summary of the regression results
print(model.summary())
```

### Comparing Species Effects Using an F-Test
To further investigate differences between species in their effect on sepal width, we conduct an F-test on the regression coefficients:



```python
# Perform an F-test to compare the coefficients of 'species[T.versicolor]' and 'species[T.virginica]'
# The vector [0, 1, -1, 0] specifies the linear hypothesis being tested:
#   - 0 for intercept (not tested)
#   - 1 for the coefficient of 'species[T.versicolor]'
#   - -1 for the coefficient of 'species[T.virginica]'
#   - 0 for 'petal_length' coefficient (not tested)
# This test checks if the difference between these two species coefficients is statistically significant
print(model.f_test([0, 1, -1, 0]))
```

### Markdown: Testing for Interaction Effects in Regression (Iris Dataset Example)
To explore whether the effect of one continuous variable on an outcome depends on another categorical variable, we include an interaction term in the regression model.

Here, we test if the relationship between petal length and sepal width varies across different species.

The model includes main effects for petal_length and species.

The interaction term (petal_length * species) tests whether the slope of petal length differs by species.

This helps uncover more nuanced, conditional relationships in the data.




```python
import statsmodels.formula.api as sm

# Fit an Ordinary Least Squares (OLS) regression with interaction term
# The formula 'sepal_width ~ petal_length + species + petal_length * species' includes:
#  - Main effects of petal length (continuous) and species (categorical)
#  - Interaction between petal length and species to check if petal length effect changes by species
result = sm.ols(formula='sepal_width ~ petal_length + species + petal_length * species', data=data).fit()

# Output the full summary of the regression results
# Including coefficients, significance levels, and model diagnostics
print(result.summary())
```
